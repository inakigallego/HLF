PRUEBAS CREAR BARCOS

```python
# 3 BARCOS DE ESLORA 2
# 2 DE ESLORA 3
# 1 ESLORA 4
import numpy as np

eslora = 3 # lo coloco de forma vertical de momento
def crear_barco(eslora):
    barco = []
    pos_ini = (np.random.randint(10), np.random.randint(10)) ; #print(pos_ini)
    #barco.append(pos_ini); print(barco)
    #barco.append((pos_ini[0]+1, pos_ini[1])); print(barco)
    barco.append(pos_ini)
    dir = np.random.choice(("V", "H"))
    p = 1
    if dir == "V":
        for pos in np.arange(eslora-1):
            barco.append((pos_ini[0]+p, pos_ini[1]))
            p = p + 1
    else:
        for pos in np.arange(eslora-1):
                barco.append((pos_ini[0], pos_ini[1]+p))
                p = p + 1  

    return(barco)
print(crear_barco(3))
```

```python
print(crear_barco(2))
print(crear_barco(3))
print(crear_barco(4))
```

CREAR FLOTA

```python
esloras_flota = [4, 3, 3, 2, 2, 2]
flota[]
for eslora in eslora_flota:
        flota.append(crear_barco(eslora))

flota
```
